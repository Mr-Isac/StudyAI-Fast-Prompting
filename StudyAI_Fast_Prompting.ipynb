# StudyAI: Asistente Inteligente para la Planificación del Estudio

## Introducción

StudyAI es una prueba de concepto que utiliza inteligencia artificial
y técnicas de Prompt Engineering para generar planes de estudio
personalizados.

La problemática abordada es la dificultad que pueden tener los
estudiantes para organizar diferentes materias, temas y tiempos
disponibles de estudio.

La POC utiliza un modelo de lenguaje para transformar la información
proporcionada por el usuario en un plan organizado de estudio.

Durante la implementación se experimentaron diferentes estrategias
de prompting, comenzando con instrucciones simples y evolucionando
hacia prompts estructurados con roles, objetivos, restricciones,
formato de salida y ejemplos.

El objetivo es demostrar cómo la construcción de prompts puede
mejorar el control y la utilidad de las respuestas generadas por
un modelo de inteligencia artificial.

## Objetivos

### Objetivo general

Desarrollar una prueba de concepto que utilice técnicas de Prompt
Engineering para generar planes de estudio personalizados mediante
un modelo de inteligencia artificial.

### Objetivos específicos

- Diseñar prompts orientados a la planificación académica.
- Experimentar con diferentes técnicas de prompting.
- Comparar los resultados obtenidos con diferentes configuraciones.
- Reducir respuestas incorrectas o contenidos inventados por el modelo.
- Permitir que el usuario proporcione sus propias materias y temas.
- Generar un plan de estudio respetando el tiempo disponible.
- Evaluar la reutilización del prompt con diferentes casos.

## Metodología

El desarrollo de la POC se realizará mediante un proceso iterativo de
diseño, prueba y mejora de prompts.

Primero se construirá un prompt inicial con instrucciones generales.
Luego se incorporarán progresivamente diferentes técnicas de Prompt
Engineering, agregando contexto, restricciones, ejemplos y una
estructura definida para la respuesta.

Cada versión será ejecutada utilizando los mismos datos para poder
observar las diferencias en los resultados.

Posteriormente, se probará el prompt optimizado con un segundo conjunto
de materias y temas. De esta manera se podrá evaluar si la solución es
capaz de reutilizar el mismo prompt con diferentes entradas.

Finalmente, se implementará una función que permita enviar el prompt
al modelo y obtener el plan de estudio generado mediante una única
consulta a la API.

El proceso seguirá las siguientes etapas:

1. Definición del problema.
2. Diseño del prompt inicial.
3. Prueba del prompt.
4. Identificación de problemas en la respuesta.
5. Optimización del prompt mediante técnicas de prompting.
6. Prueba con diferentes datos.
7. Implementación de la versión final.
8. Análisis de los resultados.

## Técnicas de prompting utilizadas

Durante la experimentación se utilizaron diferentes estrategias:

- **Prompting directo:** se proporcionaron instrucciones generales
  para generar un plan de estudio.
- **Prompt estructurado:** se dividieron las instrucciones en
  secciones como rol, objetivo, datos, reglas y formato.
- **Few-shot prompting:** se proporcionó un ejemplo de entrada y
  salida para orientar al modelo sobre el resultado esperado.
- **Restricciones:** se establecieron reglas explícitas para evitar
  que el modelo inventara materias o temas y para controlar el tiempo
  de estudio.
- **Prompt dinámico:** los datos proporcionados por el usuario se
  incorporan automáticamente al prompt mediante variables de Python.

## Experimentación con Prompts

En esta etapa se realizaron diferentes pruebas para evaluar cómo la
estructura y las instrucciones utilizadas en un prompt afectan la
respuesta del modelo.

Se comenzó con un prompt simple y posteriormente se incorporaron
progresivamente diferentes técnicas de Prompt Engineering:

1. Prompting directo.
2. Incorporación de contexto y restricciones.
3. Few-shot prompting mediante ejemplos.
4. Prompt estructurado mediante secciones.
5. Prompt dinámico utilizando los datos ingresados por el usuario.

El objetivo de la experimentación fue identificar qué configuración
permitía obtener planes de estudio más controlados, relevantes y
consistentes.

Durante las pruebas también se detectó que el modelo podía generar
contenidos que no habían sido proporcionados por el usuario. Este
comportamiento permitió identificar una limitación del prompt inicial
y justificar la incorporación de restricciones explícitas en la
versión optimizada.

```python
prompt_inicial = """
Actúa como un asistente de planificación de estudios.

Crea un plan de estudio para un estudiante que tiene que preparar
varias materias antes de sus exámenes.

El plan debe organizar las materias y temas de forma clara,
indicando qué estudiar cada día y cuánto tiempo dedicar.
"""

respuesta = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {"role": "user", "content": prompt_inicial}
    ]
)

print(respuesta.choices[0].message.content)
```

```python
prompt_mejorado = """
Actúa como un asistente especializado en planificación académica.

Crea un plan de estudio personalizado utilizando estos datos:

Materias:
- Matemáticas: ecuaciones y funciones.
- Historia: Revolución Francesa.
- Inglés: tiempos verbales.

Fecha del examen: dentro de 7 días.
Tiempo disponible: 2 horas por día.

Organiza el plan de lunes a domingo.
Para cada día indica:
- Materia.
- Tema.
- Tiempo de estudio.
- Actividad recomendada.

No asignes más de 2 horas de estudio por día.
Prioriza los temas más importantes y distribuye la carga de manera equilibrada.
"""

respuesta = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {"role": "user", "content": prompt_mejorado}
    ]
)

print(respuesta.choices[0].message.content)
```

```python
prompt_few_shot = """
Actúa como un asistente especializado en planificación académica.

Tu tarea es crear planes de estudio personalizados.

Ejemplo:

Entrada:
Materia: Matemáticas
Tema: Ecuaciones
Tiempo disponible: 1 hora

Salida esperada:
Materia: Matemáticas
Tema: Ecuaciones
Duración: 1 hora
Actividad: Resolver ejercicios prácticos y revisar los errores.

Ahora crea un plan utilizando estos datos:

Materias:
- Matemáticas: ecuaciones y funciones.
- Historia: Revolución Francesa.
- Inglés: tiempos verbales.

Fecha del examen: dentro de 7 días.
Tiempo disponible: 2 horas por día.

Organiza el plan de lunes a domingo.
Para cada día indica:
- Materia.
- Tema.
- Tiempo de estudio.
- Actividad recomendada.

No asignes más de 2 horas de estudio por día.
Distribuye la carga de manera equilibrada.
"""

respuesta = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {"role": "user", "content": prompt_few_shot}
    ]
)

print(respuesta.choices[0].message.content)
```

```python
prompt_controlado = f"""
# ROL
Eres un asistente experto en planificación académica.

# OBJETIVO
Crear un plan de estudio utilizando EXCLUSIVAMENTE las materias y
contenidos proporcionados por el estudiante.

# DATOS DEL ESTUDIANTE
Materias y contenidos: {materias}
Tiempo disponible por día: {tiempo}
Días disponibles: {dias}

# REGLAS
- NO inventes materias ni temas que el estudiante no haya proporcionado.
- NO agregues contenidos externos.
- Distribuye únicamente los contenidos recibidos.
- No superes el tiempo disponible por día.
- Distribuye la carga de estudio de manera equilibrada.
- Si falta información sobre los temas, indícalo en lugar de inventarlos.

# FORMATO
Para cada día indica:
- Materia
- Tema
- Duración
- Actividad recomendada

Genera únicamente el plan de estudio.
"""

plan_controlado = generar_plan(prompt_controlado)

print("\nPLAN DE ESTUDIO\n")
print(plan_controlado)
```

## Comparación de prompts

Se probaron diferentes configuraciones de prompts para evaluar cómo las
instrucciones afectan la calidad de la respuesta generada.

- **Prompt inicial:** proporciona instrucciones generales, pero deja varios
  aspectos de la respuesta a criterio del modelo.
- **Prompt mejorado:** incorpora información concreta, restricciones y un
  formato esperado.
- **Few-shot prompting:** incorpora un ejemplo de entrada y salida para
  orientar al modelo sobre el formato esperado.

La comparación permite observar cómo agregar contexto, restricciones y
ejemplos puede producir respuestas más consistentes y adecuadas para
la planificación de estudios.

## Optimización del prompt

La primera versión del prompt permitía que el modelo incorporara
contenidos que no habían sido proporcionados por el usuario.

Para solucionar este problema se incorporaron instrucciones más
específicas, incluyendo:

- Uso exclusivo de las materias y temas proporcionados.
- Prohibición de inventar contenidos.
- Restricción del tiempo máximo de estudio.
- Definición de un formato de respuesta.
- Distribución equilibrada de las sesiones.

Como resultado, el modelo genera un plan más controlado y alineado
con la información ingresada por el estudiante.


## POC Interactiva

En esta etapa se incorpora una interacción básica con el usuario para
demostrar el funcionamiento de la solución en un escenario más cercano
a su uso real.

El usuario puede ingresar sus propias materias y temas, indicar cuánto
tiempo tiene disponible por día y especificar durante cuántos días
necesita organizar su estudio.

Estos datos se incorporan dinámicamente al prompt estructurado y se
envían al modelo de inteligencia artificial para generar un plan de
estudio personalizado.

De esta manera, la POC no depende de un conjunto de datos fijo, sino
que puede generar diferentes planes a partir de las necesidades de
cada usuario.

### Flujo de funcionamiento

1. El usuario ingresa sus materias y temas.
2. Indica el tiempo disponible por día.
3. Indica la cantidad de días disponibles.
4. Los datos se incorporan al prompt.
5. El modelo procesa la información.
6. Se genera y muestra el plan de estudio.

```python
materias = input("¿Qué materias necesitas estudiar? ")
tiempo = input("¿Cuánto tiempo tienes disponible por día? ")
dias = input("¿Cuántos días tienes para estudiar? ")

prompt_usuario = f"""
Actúa como un asistente experto en planificación académica.

Crea un plan de estudio personalizado utilizando estos datos:

Materias: {materias}
Tiempo disponible por día: {tiempo}
Días disponibles: {dias}

Distribuye los contenidos de manera equilibrada.
No superes el tiempo disponible por día.

Para cada día indica:
- Materia
- Tema
- Duración
- Actividad recomendada

Genera únicamente el plan de estudio.
"""

plan_usuario = generar_plan(prompt_usuario)

print("\nPLAN DE ESTUDIO\n")
print(plan_usuario)
```

## Optimización final del prompt

A partir de las pruebas anteriores se realizó una última optimización
del prompt para aumentar el control sobre las respuestas generadas.

Se incorporaron restricciones adicionales para asegurar que cada sesión
del plan estuviera relacionada directamente con los temas proporcionados
por el usuario y evitar que el modelo agregara contenidos innecesarios.

La versión final utiliza una estructura clara dividida en:

- Rol del modelo.
- Objetivo.
- Datos del estudiante.
- Reglas y restricciones.
- Formato esperado de la respuesta.

Esta estructura permite reducir la libertad innecesaria del modelo y
obtener resultados más consistentes y alineados con la información
proporcionada.

La versión optimizada será utilizada posteriormente para realizar una
prueba con un segundo conjunto de datos y comprobar su reutilización.

```python
materias = input(
    "Ingresa las materias y sus temas "
    "(ejemplo: Matemáticas: ecuaciones y funciones; "
    "Historia: Revolución Francesa; Inglés: tiempos verbales): "
)

tiempo = input("¿Cuánto tiempo tienes disponible por día? ")
dias = input("¿Cuántos días tienes para estudiar? ")

prompt_usuario = f"""
# ROL
Eres un asistente experto en planificación académica.

# OBJETIVO
Crear un plan de estudio personalizado utilizando exclusivamente
las materias y temas proporcionados por el estudiante.

# DATOS DEL ESTUDIANTE
Materias y temas: {materias}
Tiempo disponible por día: {tiempo}
Días disponibles: {dias}

# REGLAS
- No inventes materias ni temas.
- Utiliza únicamente los contenidos proporcionados.
- No superes el tiempo disponible por día.
- Distribuye la carga de estudio de manera equilibrada.
- Cada sesión debe estar relacionada directamente con un tema proporcionado.
- Incluye actividades apropiadas para cada tema.
- No agregues información académica que no sea necesaria para organizar el estudio.

# FORMATO
Para cada día indica:
- Materia
- Tema
- Duración
- Actividad recomendada

Genera únicamente el plan de estudio.
"""

plan_usuario = generar_plan(prompt_usuario)

print("\nPLAN DE ESTUDIO\n")
print(plan_usuario)
```
